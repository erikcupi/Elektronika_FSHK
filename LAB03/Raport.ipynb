## Qëllimi i eksperimentit

Qëllimi i kësaj pune është të mësojmë përdorimin e oshiloskopit, të matim tensionin dhe periodën e një sinjali dhe të shohim lidhjen ndërmjet periodës dhe frekuencës për matje me kondensatorë të ndryshëm.

## Pak teori

Oshiloskopi na lejon të shohim ndryshimin e tensionit në varësi të kohës. Në boshtin vertikal lexohet tensioni, ndërsa në boshtin horizontal koha.

Për të gjetur tensionin dhe periodën përdorim numrin e ndarjeve në ekran dhe shkallën përkatëse:

$$U=n_y\cdot (V/div),\qquad T=n_x\cdot (s/div)$$

Frekuenca gjendet nga:

$$f=\frac{1}{T}$$

Qarku RC përbëhet nga një rezistencë dhe një kondensator. Kondensatori ngarkohet dhe shkarkohet gjatë punës së qarkut. Konstantja e kohës është $\tau=RC$, ku $R$ matet në om dhe $C$ në farad. Konstantja e kohës nuk është domosdoshmërisht e njëjtë me periodën e sinjalit.

## Eksperimenti 1 – Matje me oshiloskop

Nga ndarjet e lexuara në ekran morëm këto të dhëna:

| Madhësia | Numri i ndarjeve | Shkalla | Rezultati |
| :--- | :---: | :---: | :---: |
| Tensioni | 2 | 2 V/div | 4 V |
| Perioda | 4 | 0,2 ms/div | 0,8 ms |

### Llogaritjet

$$U=2\cdot 2=4\ V$$

$$T=4\cdot 0.2=0.8\ ms=0.0008\ s$$

$$f=\frac{1}{0.0008}=1250\ Hz=1.25\ kHz$$

Nga matja dolën tensioni 4 V dhe frekuenca 1250 Hz. Vlera e tensionit i përket dy ndarjeve të matura; pa pamjen e sinjalit nuk mund të përcaktojmë nëse është amplitudë apo tension kulm më kulm.

## Eksperimenti 2 – Qarqet RC

Në pjesën e dytë janë shënuar perioda dhe frekuenca për katër kondensatorë.

### Matjet eksperimentale

| Kondensatori | T (s) | Frekuenca e shënuar | Frekuenca nga $1/T$ |
| :---: | :---: | :---: | :---: |
| 1 | $3\cdot10^{-6}$ | 165,6 kHz | 333,33 kHz |
| 2 | $6\cdot10^{-6}$ | 337,4 kHz | 166,67 kHz |
| 3 | $2\cdot10^{-6}$ | 510 kHz | 500 kHz |
| 4 | $1\cdot10^{-6}$ | 1 kHz | 1000 kHz |

### Përpunimi i matjeve

$$f_1=\frac{1}{3\cdot10^{-6}}=333.33\ kHz$$

$$f_2=\frac{1}{6\cdot10^{-6}}=166.67\ kHz$$

$$f_3=\frac{1}{2\cdot10^{-6}}=500\ kHz$$

$$f_4=\frac{1}{1\cdot10^{-6}}=1000\ kHz=1\ MHz$$

Kur krahasojmë rezultatet, vërejmë se frekuencat e dy matjeve të para mund të jenë shkëmbyer gjatë shënimit. Te matja e tretë ndryshimi është i vogël, ndërsa te e katërta ka mundësi të jetë shënuar kHz në vend të MHz. Vlerat origjinale janë lënë në tabelë për t'u kontrolluar.

## Përfundimi

Gjatë kësaj pune pamë se si mund të përcaktojmë tensionin, periodën dhe frekuencën duke përdorur oshiloskopin. Nga llogaritjet e eksperimentit të dytë kuptuam se sa më e vogël të jetë perioda, aq më e madhe është frekuenca. Disa nga vlerat e shënuara nuk përputhen me llogaritjet, prandaj duhen kontrolluar përsëri matjet dhe njësitë. Pa vlerat e rezistencës dhe kapacitetit nuk mund të llogarisim konstanten RC.
