#  Përmbledhje: Çfarë Mësuam rreth Markdown në JupyterLite

## 1. Çfarë është Markdown?
Markdown është një gjuhë shënimi e lehtë që përdoret për të formatuar tekstin në JupyterLite përmes simboleve të thjeshta (`#`, `*`, `|`, etj.), pa pasur nevojë për HTML.

## 2. Pse na duhet në Jupyter?
* **Strukturim:** Ndihmon në organizimin e projektit në seksione të qarta.
* **Dokumentim:** Lejon shpjegimin e logjikës prapa kodit (Python, R, etj.).
* **Profesionalizëm:** E kthen notebook-un në një raport ose ditar të lexueshëm.

---

## 3. Elementët Kryesorë që Përdorëm:

* **Titujt:** `# H1`, `## H2`, `### H3`
* **Formatimi:** 
  * **Bold:** `**tekst**`
  * *Italic:* `*tekst*`
  * `Inline Code`: `` `kodi` ``
  * ~~Strikethrough~~: `~~tekst~~`
* **Listat:** 
  * Me pika (`*`)
  * Të numëruara (`1.`)
* **Tabelat dhe Lidhjet:** Për organizimin e të dhënave dhe vendosjen e linkeve `[Faqja](url)`.
* **LaTeX (Matematikë):** `$E = mc^2$` për brenda rreshtit dhe `$$ ... $$` për formula të veçanta.
* **Blloqet e Shënimeve:** Përdorimi i `>` për të krijuar kuti vëmendjeje ose këshilla.
